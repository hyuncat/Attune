# COCO-E2: generation and exact evaluation tracks

COCO-E2 injects symbolic mistakes into isolated CocoChorales MIDI stems and renders them with the original General MIDI instrument. The implementation is `experiments/modules/mistake/datasets/MistakeCases.py`; `MistakeInjector.py` implements the edits. Cases use equal weights for the five edit types, rate 0.25, relative duration factors 0.5–1.5 with a 0.3-second minimum error, and no timing or duration jitter. Rate 0 provides clean controls.

`benchmark-tracks.json` freezes 260 source stems, seed 0, and clean/injected conditions (520 logical performances per method). The saved benchmark has 532 asset variants across methods; the manifest retains each variant and records which methods used it. `sweeps-tracks.json` freezes the revised 195-stem tuning cohort, with clean seed 0 and injected seeds 0 and 1 (585 performances; 390 injected). It includes the frozen horn replacement. These are separate selections, not newly sampled tracks.

Restore the original CocoChorales materialized stems under `../coco/cocochorales_tiny/` and saved experiment assets under `../../results/caches/`. The two cells below copy exact score, performed MIDI, audio, and truth files after SHA-256 verification; no detector or benchmark runs. Each manifest records the original generation specification and file hashes. Set `CHECK_ONLY=True` to verify without copying.

For missing generated assets, `--regenerate` on the script calls the production generator using the frozen score, seed and rate. It still requires matching hashes: rendering libraries, soundfont, generator, and annotation formatting must match the historical run. If they differ, restore the frozen assets to obtain byte-identical evaluation data. The script never silently substitutes a newly sampled or changed performance.
