# Volume-reference ablation: percentile versus maximum

Can the live and post-hoc volume rules be unified without sacrificing pitch/voicing accuracy?
The current rules are not algebraically equivalent: equality requires
$\alpha_{\mathrm{live}}=\alpha_{\mathrm{post}}Q_{0.95}(\mathrm{RMS})/\max(\mathrm{RMS})$,
and that ratio varies by recording. Live mode uses a **running** maximum, not a future-aware recording-wide maximum.

Paired variants on the frozen 449-track cohort (260 CocoChorales, 149 URMP, 40 Bach10):

| Variant | Volume threshold | Purpose |
|---|---|---|
| `current_p95` | post-hoc alpha × recording p95 | Current production control |
| `global_max_live_alpha` | live alpha × recording max | Proposed shared coefficient/reference type |
| `running_max_live_alpha` | live alpha × running max | Reuse the actual causal live volume rule |
| `global_max_post_alpha` | post-hoc alpha × recording max | Isolate the reference change at fixed alpha |

Only the volume rule changes. HMM settings, candidates, confidence thresholds, range limits, and scorer stay fixed. The volume rule affects **both HMM tracking regions and final voicing**, so every variant reruns smoothing from the same cached raw candidates. Final-gated smoothed caches cannot recover rejected pitches or reconstruct a changed HMM path; they are therefore not valid inputs for this comparison. Raw live-gated values are safe to reuse because the HMM uses the retained candidates and probabilities.

No pitch detection runs and production caches are never written. Missing/incompatible raw caches stop the run. Per-track ablation scores are cached separately with hashes of raw cache contents, reference, configuration and relevant source files; reruns reuse these results. Historical final-voicing-sweep arrays lack RMS and contain a fixed tracking mask, so they cannot support this ablation.

Metrics use the existing 10 ms frame scorer: pooled and median per-recording accuracy, voicing recall/false alarms, and paired accuracy changes. Diagnostics report volume-mask disagreement and changed final outputs. These are exploratory results on the existing tuning cohort, not held-out validation or a live-latency benchmark. Use the `at-venv` kernel.

## Live volume gate with a running 95th percentile

For capture frame $i$, compare its RMS with $\alpha Q_{0.95}(\mathrm{RMS}_1,\ldots,\mathrm{RMS}_i)$. The reference includes all capture frames observed so far, including silence and the current frame. The distribution starts fresh for each recording. The experiment crosses running maximum and running p95 with coefficients 0.03 and 0.04. Running maximum at 0.03 is the production control. Running p95 at 0.04 shares the post-hoc percentile and coefficient.

All four variants use the same unsmoothed live pitch candidates and the live periodicity threshold of 0.985. We recover the detector’s pre-volume candidate decision from the raw cache, drop the padded boundary frames, and score the complete capture windows on the live benchmark’s time grid. For each track, the production control must reproduce every saved live scoring count before its comparison is accepted. The tables measure accuracy on the existing 449-recording cohort. Real-time update latency would require timing an implementation in the live detector.

With the same **0.04** coefficient used post-hoc, running p95 raised pooled live overall accuracy from **93.292% to 93.357%**. Voicing recall rose by **0.259 percentage points**, and the false-alarm rate rose by **0.734 points**. The pooled accuracy change was +0.068 points on CocoChorales, +0.069 on URMP, and −0.014 on Bach10. At 0.03, running p95 achieved nearly the same pooled accuracy (93.352%) with a higher false-alarm rate (14.032%).

These measurements support a shared volume rule on this cohort if the increase in false alarms is acceptable. In that formulation both modes use 4% of the 95th-percentile RMS. Live mode takes the percentile over frames received through the current frame, while post-hoc mode takes it over the completed recording.
